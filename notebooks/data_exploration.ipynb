### Data Exploration
 In this notebook, i will examine the dataset structure, date ranges, missing values, duplicate records, and key variables of both datasets to assess their suitability for analysis. 


```python
import pandas as pd
import matplotlib as plt
import seaborn as sns

# load data
billboard = pd.read_csv("/Users/apeksha/Desktop/Bootcamp/Final Project/data/raw/hot100.csv")
tiktok_trends = pd.read_csv("/Users/apeksha/Desktop/Bootcamp/Final Project/data/raw/trending_music.csv")
```

##### investigating "last week"

```python
billboard["Last Week"].value_counts(dropna=False).head(20)
```

```python
last_week_numeric = pd.to_numeric(billboard["Last Week"], errors="coerce")

print("Non-numeric values:")
print(billboard.loc[last_week_numeric.isna(), "Last Week"].value_counts(dropna=False).head(20))

print("Number of non-numeric values:", last_week_numeric.isna().sum())
```

```python
print("Minimum rank:", billboard["Rank"].min())
print("Maximum rank:", billboard["Rank"].max())
print("Unique ranks:", billboard["Rank"].nunique())

print("\nFrequency of each rank:")
print(billboard["Rank"].value_counts().sort_index())
```

```python
print("Earliest date:", billboard["Date"].min())
print("Latest date:", billboard["Date"].max())

print("Number of unique chart dates:", billboard["Date"].nunique())

print("\nNumber of rows per chart date:")
print(billboard.groupby("Date").size().describe())
```

```python
# display records with non-numeric Weeks in Charts
unusual_weeks = billboard[
    pd.to_numeric(
        billboard["Weeks in Charts"], errors="coerce"
    ).isna()
]

display(unusual_weeks.head(20))
```

```python
unusual_weeks = billboard[billboard["Weeks in Charts"].astype(str) == "-"]

print("Number of records:", len(unusual_weeks))
print("Dates covered:")
print(unusual_weeks["Date"].value_counts().sort_index())

display(unusual_weeks[["Date", "Song", "Artist", "Rank","Last Week", "Weeks in Charts"]].head(10))
```

```python
# convert date to a proper datetime format
tiktok_trends["date"] = pd.to_datetime(tiktok_trends["date"], errors="coerce")

print("Earliest date:", tiktok_trends["date"].min())
print("Latest date:", tiktok_trends["date"].max())
```

```python
# missing values
print("Missing values per column:")
display(tiktok_trends.isnull().sum().to_frame("missing_values"))

# duplicate rows
print("Exact duplicate rows:", tiktok_trends.duplicated().sum())
```

```python
tiktok_trends["window_days"].value_counts(dropna=False)
```

```python
tiktok_trends["country_code"].value_counts(dropna=False)
```

```python
# exploring rank columns and the datas in it
print("Minimum rank:", tiktok_trends["rank"].min())
print("Maximum rank:", tiktok_trends["rank"].max())
print("Unique ranks:", tiktok_trends["rank"].nunique())

print("\nFrequency of each rank:")
print(tiktok_trends["rank"].value_counts().sort_index())
```

```python
# exploring unoque songs, clips and freq appearing songs
print("Unique Song id:", tiktok_trends["song_id"].nunique())
print("Unique clip id:", tiktok_trends["clip_id"].nunique())
print("Unique song title:", tiktok_trends["music_title"].nunique())

print("\nFrequently appearing songs:")
print(tiktok_trends["music_title"].value_counts().head(10))
```

```python
# explore rank_diff to understand missing values
print(tiktok_trends["rank_diff_label"].value_counts(dropna=False))
```

```python
print(tiktok_trends["rank_diff_type"].value_counts(dropna=False))
```

```python
# inspect records with missing rank_diff
missing_rank_diff = tiktok_trends[tiktok_trends["rank_diff"].isna()]

print("Rows with missing rank_diff:", len(missing_rank_diff))

display(
    missing_rank_diff[
        [
            "date",
            "window_days",
            "rank",
            "music_title",
            "music_author",
            "rank_diff",
            "rank_diff_type",
            "rank_diff_label",
        ]
    ].head(15)
)
```

```python
duplicate_song_records = tiktok_trends.duplicated(
    subset=["date", "window_days", "song_id"],
    keep=False
)

print("Rows involved in repeated song records:",duplicate_song_records.sum())

display(
    tiktok_trends.loc[
        duplicate_song_records,
        ["date", "window_days", "song_id",
         "music_title", "rank", "clip_id"]
    ].head(20)
)
```

```python
tiktok_7day = tiktok_trends[tiktok_trends["window_days"] == 7]

print("7-day dataset shape:", tiktok_7day.shape)
print("Unique songs:", tiktok_7day["song_id"].nunique())

print("\nRank statistics:")
display(tiktok_7day["rank"].describe())

print("\nRecords per date:")
display(tiktok_7day.groupby("date").size().describe())
```

```python

# Check the date ranges of both datasets
print("Billboard date range:")
print(billboard["Date"].min(), "to", billboard["Date"].max())

print("\nTikTok 7-day date range:")
print(tiktok_7day["date"].min(), "to", tiktok_7day["date"].max())

# Compare the overall date ranges
overlap_start = max(billboard["Date"].min(),tiktok_7day["date"].min())

overlap_end = min(billboard["Date"].max(),tiktok_7day["date"].max())

print("\nOverlapping period:")
print(overlap_start, "to", overlap_end)

# Count Billboard records within that period
billboard_overlap = billboard[billboard["Date"].between(overlap_start, overlap_end)]

print("\nBillboard rows during the overlapping period:",len(billboard_overlap))

print("Billboard chart dates during the overlapping period:",billboard_overlap["Date"].nunique())
```
